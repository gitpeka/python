```python
#array/list, tipe data yang memiliki banyk data lainnya
bilangan_bulat = [10,20,30]

print(bilangan_bulat)

#mengambil bilangan berdasarkan  index
print(bilangan_bulat[0])

print(bilangan_bulat[1])

#menambahkan data diakhir, append
bilangan_bulat.append(50)
print(bilangan_bulat)

#menambahkan data berdasarkan lokasi, insert
bilangan_bulat.insert(0,12)
print(bilangan_bulat)

bilangan_bulat.insert(2,2)
bilangan_bulat.insert(3,50)
print(bilangan_bulat)

#pop (menghilangkan berdasarkan indekx)
bilangan_bulat.pop(0)
print(bilangan_bulat)



#tugas
orang = []
orang.append('wista')
print(orang)
orang.append('ethan')
print(orang)
orang.append('davin')
print(orang)
orang.append('mulya')
print(orang)
orang.append('wenwen')
print(orang)
orang.append('eillan')
print(orang)
orang.append('ham')
print(orang)
orang.append('chiara')
print(orang)
orang.pop(0)
print(orang)




index1 = [2,4,6]
index2 = [3,5,7]

hasil = index1[0] * index2[0]
print(hasil)
hasil = index1[1] * index2[1]
print(hasil)
hasil = index1[2] * index2[2]
print(hasil)
```

```python
#untuk mencari tipe data menggunakan pandas series
import pandas as pd

mydata = {
    "name" : ["rand","rand"],
    "value" : [10,20]
}

myseries = pd.Series(mydata)
print(myseries)

mynumber = [10,20,30,40,50]
print(pd.Series(mynumber))

mynrealnumber = [0.1,0.2,0.3]
print(pd.Series(mynrealnumber))
```
